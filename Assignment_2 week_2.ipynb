```python
import tensorflow as tf

from tensorflow import keras 

from tensorflow.keras import layers

from tensorflow.keras.models import Sequential
```

## Load the Data Directory

```python
data_dir = r"C:\Users\hp\Desktop\DSML files\FishImgDataset"
```

```python
test_data = r"C:\Users\hp\Desktop\DSML files\FishImgDataset\test"
```

```python
train_data = r"C:\Users\hp\Desktop\DSML files\FishImgDataset\train"
```

```python
img_height = 200
img_width = 200
batch_size = 40
epochs = 100
```

```python
test_ds = tf.keras.utils.image_dataset_from_directory(
    test_data,
    labels = 'inferred',
    label_mode = 'categorical',
    seed = 123,
    image_size = (img_height,img_width),
    batch_size = batch_size
)
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    train_data,
    labels = 'inferred',
    label_mode = 'categorical',
    seed = 123,
    image_size = (img_height,img_width),
    batch_size = batch_size
)
```

## Model Building

#### First (input_layer)

```python
model = Sequential([
    # input layer
    layers.Conv2D(30,(3,3),padding= "same",activation="relu",input_shape = (img_height, img_width, 3)),
    
    # Hidden Layer 1
    layers.Conv2D(50, (3,3), padding="same", activation="relu"),
    layers.MaxPooling2D(),
    layers.Dropout(0.3),
    # Hidden Layer 2
    layers.Conv2D(70, (3,3), padding="same", activation="relu"),
    layers.MaxPooling2D(),
    layers.Dropout(0.4),
    # Hidden Layer 3
    layers.Conv2D(90, (3,3), padding="same", activation="relu"),
    layers.MaxPooling2D(),
    layers.Dropout(0.4),
    # Hidden Layer 4
    layers.Conv2D(100, (3,3), padding="same", activation="relu"),
    layers.MaxPooling2D(),
    layers.Dropout(0.5),
    # Hidden Layer 5
    layers.Conv2D(100, (3,3), padding="same", activation="relu"),
    layers.AveragePooling2D(2,2),
    layers.Dropout(0.5),
    
    # Output Layer
    layers.Flatten(),
    layers.Dense(100, activation="relu"),
    
    layers.Dense(31, activation= 'softmax'),
    # Note Activation function softmax was used cause our model have 31 classes to make it's predictions from
])
```

```python
model.summary()
```

### Model Compiling

```python
model.compile(optimizer=  tf.keras.optimizers.Adam(),
              loss = tf.keras.losses.CategoricalCrossentropy(from_logits = False),
              metrics = ['accuracy'])
```

## Model Training

```python
history = model.fit(
    train_ds,
    
    validation_data = test_ds,
    epochs = epochs,
)
```
